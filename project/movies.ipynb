```python
import ast

meta_data["genres"] = meta_data["genres"].apply(
    lambda x: [i["name"] for i in ast.literal_eval(x)]
)
meta_data["genres"] = meta_data["genres"].apply(lambda x: " ".join(x))
meta_data["genres"] = meta_data["genres"].fillna("").apply(lambda x: x.split())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

overview_vector = TfidfVectorizer(stop_words='english', max_features=5000)
overview_tfidf = overview_vector.fit_transform(meta_data["overview"])
tagline_vector = TfidfVectorizer()
tagline_tfidf = tagline_vector.fit_transform(meta_data["tagline"])
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
genres_vec = mlb.fit_transform(meta_data["genres"])
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
numeric_matrix = scaler.fit_transform(meta_data[['runtime', 'popularity']].fillna(0))
```

```python
from sklearn.preprocessing import OneHotEncoder

lang_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)

language_vec = lang_encoder.fit_transform(
    meta_data[["original_language"]].fillna("")
)
```

```python
import numpy as np
from scipy.sparse import hstack

combined_features = hstack([
    overview_tfidf,   
    genres_vec,
    language_vec,
    numeric_matrix
])
```

PCA

```python
from sklearn.decomposition import PCA

combined_dense = combined_features.toarray()

pca = PCA(n_components=50, random_state=42)

pca_features = pca.fit_transform(combined_dense)
print(pca_features.shape)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.7,
    min_samples=5
)

clusters = dbscan.fit_predict(pca_features)

meta_data["cluster"] = clusters
```

```python
print(meta_data["cluster"].value_counts())
```

```python
def recommend_cluster(movie_title, n=10):

    idx = meta_data[meta_data["title"] == movie_title].index[0]

    cluster = meta_data.loc[idx, "cluster"]

    recommendations = meta_data[
        (meta_data["cluster"] == cluster) &
        (meta_data.index != idx)
    ][[
        "title",
        "genres",
        "poster_path",
        "vote_average",
        "release_date"
    ]].head(n)

    return recommendations
```
