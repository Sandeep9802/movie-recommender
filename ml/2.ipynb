```python
df=pd.get_dummies(df,columns=["model"])
```

```python
from sklearn.preprocessing import LabelEncoder
li=LabelEncoder()
df["transmission"]=li.fit_transform(df["transmission"])
df["fuelType"]=li.fit_transform(df["fuelType"])
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler

s = StandardScaler()

df["mileage"] = s.fit_transform(df[["mileage"]])
df["tax"] = s.fit_transform(df[["tax"]])
df["mpg"] = s.fit_transform(df[["mpg"]])

```

```python
df
```

```python
x = df.drop(columns=["price"])
y=df["price"]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.metrics import r2_score
model=LinearRegression()
model.fit(x_train,y_train)
pred = model.predict(x_test)
r2=r2_score(y_test,pred)
# Adjusted R2
n = x_test.shape[0]   # total rows
p = x_test.shape[1]   # total features

adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))
print(r2)
print("Adjusted R2:", adjusted_r2)
```
