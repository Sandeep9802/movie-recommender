```python
df.columns
df.shape
```

```python
df["high_ai_influence"] = (df["algorithmic_content_exposure"]*df["content_personalization_intensity"])
```

```python
df.head(5)
```

```python
df["manupation_score"]=(df["echo_chamber_score"]+df["emotional_manipulation_index"])
df["video_platform_user"]=(df["platform_YouTube"]+df["platform_TikTok"]+df["platform_Twitch"])
df["Social_Media"]=(df["platform_Facebook"]+df["platform_Instagram"]+df["platform_Snapchat"]+df["platform_X/Twitter"])
df["Community"]=df["platform_Reddit"]+df["platform_Discord"]
df['social_engagement'] = (df['Social_Media']+ df['Community'])
df['ai_influence_score'] = (df['algorithmic_content_exposure']+ df['content_personalization_intensity']+ df['ai_addiction_probability'])
df['filter_bubble_score'] = (df['echo_chamber_score']* df['content_personalization_intensity'])
```

```python
drop_cols = [
    'echo_chamber_score',
    'Social_Media',
    'Community',
    'algorithmic_content_exposure',
    'content_personalization_intensity',
     'year',

    # country columns
    'country_Brazil',
    'country_Canada',
    'country_China',
    'country_France',
    'country_Germany',
    'country_India',
    'country_Indonesia',
    'country_Italy',
    'country_Japan',
    'country_Mexico',
    'country_Nigeria',
    'country_Pakistan',
    'country_Saudi Arabia',
    'country_South Korea',
    'country_Spain',
    'country_Turkey',
    'country_UAE',
    'country_UK',
    'country_USA',

    # age group
    'age_group_Children',
    'age_group_Senior',
    'age_group_Teen',
    'age_group_Young Adult',

    # gender
    'gender_Male',
    'gender_Other'
]
df.drop(columns=drop_cols, inplace=True)
df.drop(columns=["platform_YouTube","platform_TikTok","platform_Twitch","platform_Facebook","platform_Instagram",
                 "platform_Snapchat","platform_X/Twitter","platform_Reddit","platform_Discord"],inplace=True)
df.drop(columns=["user_id"], inplace=True)
df.drop(columns=["platform_LinkedIn"], inplace=True)
```

```python
sc=StandardScaler()
df[["filter_bubble_score","high_ai_influence","ai_influence_score"]]=sc.fit_transform(df[["filter_bubble_score","ai_influence_score","high_ai_influence"]])
```

```python
from scipy.stats import pearsonr

target = "high_ai_influence"

for col in df.columns:
    
    if col != target:
        
        corr, p_value = pearsonr(df[col], df[target])
        
        print(f"{col} --> Correlation: {corr:.3f}")
```

```python
df
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# create 4 bins
df['ai_influence_bin'] = pd.qcut(
    df['high_ai_influence'],
    q=4,
    labels=False
)
cat_features = [
    'video_platform_user',
    'social_engagement'
]


chi2_results = {}

for col in cat_features:

    contingency = pd.crosstab(
        df[col],
        df['ai_influence_bin']
    )

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Accept Null (Drop Feature)'
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

# convert into dataframe
chi2_df = pd.DataFrame(chi2_results).T

print(chi2_df)
```

```python
df.drop(columns=["video_platform_user","ai_addiction_probability","ai_influence_bin"],inplace=True)
df.drop(columns=["social_engagement"],inplace=True)
```

```python
x=df[["emotional_manipulation_index","high_ai_influence","ai_influence_score","filter_bubble_score"]]
y=df["manupation_score"]
```

```python
from sklearn.model_selection import train_test_split
X_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error

from xgboost import XGBRegressor

xgb = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

xgb.fit(X_train, y_train)

pred = xgb.predict(x_test)

```

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    root_mean_squared_error
)

# Metrics
r2 = r2_score(y_test, pred)
mae = mean_absolute_error(y_test, pred)
rmse = root_mean_squared_error(y_test, pred)

print("R2 Score:", r2)
print("MAE:", mae)
print("RMSE:", rmse)

# Adjusted R2
n = x_test.shape[0]   # total rows
p = x_test.shape[1]   # total features

adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))

print("Adjusted R2:", adjusted_r2)
```
