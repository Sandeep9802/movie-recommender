```python
import pandas as pd
from sklearn.metrics import accuracy_score,classification_report
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
df
```

```python
df.columns
```

```python
df.drop(columns=["PassengerId","Cabin"],inplace=True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df
```

```python
df.drop(columns="Ticket",inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
li=LabelEncoder()
df["Sex"]=li.fit_transform(df["Sex"])
df["Embarked"]=li.fit_transform(df["Embarked"])
```

```python
df
```

```python
df.drop(columns="Name",inplace=True)
```

```python
df
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
x=df.drop("Survived",axis=1)
y=df["Survived"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
pred=model.predict(x_test)
print(accuracy_score(y_test,pred))
print(classification_report(y_test,pred))
```

```python
x_test
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=7)
knn.fit(x_train,y_train)
pred_1=knn.predict(x_test)
print(accuracy_score(y_test,pred_1))
print(classification_report(y_test,pred_1))
```

```python
from sklearn.naive_bayes import GaussianNB
bayes=GaussianNB()
bayes.fit(x_train,y_train)
pred_2=bayes.predict(x_test)
print(accuracy_score(y_test,pred_2))
print(classification_report(y_test,pred_2))
```

```python
from sklearn.tree import DecisionTreeClassifier
tree=DecisionTreeClassifier(random_state=42)
tree.fit(x_train,y_train)
pred_3=tree.predict(x_test)
print(accuracy_score(y_test,pred_3))
print(classification_report(y_test,pred_3))
```

```python
from sklearn.svm import SVC
svm=SVC(kernel='rbf')
svm.fit(x_train,y_train)
pred_4=svm.predict(x_test)
print(accuracy_score(y_test,pred_4))
print(classification_report(y_test,pred_4))
```

```python
from sklearn.model_selection import cross_val_score
sc=StandardScaler()
x_sclaced=sc.fit_transform(x)
modell=cross_val_score(svm,x_sclaced,y,cv=5,scoring="accuracy")
print(modell)
print(modell.mean())
```
