```python
from sklearn.feature_extraction.text import TfidfVectorizer
documents=[
    "i love you",
    "pizza is the best",
   " i love pasta"
   "pasta is great"
]
vector=TfidfVectorizer()
x=vector.fit_transform(documents)
print(vector.get_feature_names_out())# volcabulary
print(x.toarray())#matrix
```
